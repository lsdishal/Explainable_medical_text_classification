## Explainable Medical Text Classification using Transformer Models

This notebook downloads the **Symptom2Disease** dataset, performs EDA on disease class distributions and text length, and saves train/val/test splits.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.downloader import download_symptom2disease, prepare_splits
from src.utils.config import get_config

# Load configuration
cfg = get_config('../config.yaml')
print(f"Project: {cfg.project.name}")
```

```python
# Download and load dataset
df = download_symptom2disease('../config.yaml')
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Plot Disease Category Distribution
plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='label', order=df['label'].value_counts().index, palette='crest')
plt.title('Symptom2Disease Class Distribution')
plt.xlabel('Sample Count')
plt.ylabel('Disease Category')
plt.tight_layout()
plt.show()
```

```python
# Generate train/val/test splits
train_df, val_df, test_df, label2id, id2label = prepare_splits('../config.yaml')
print(f"Train size: {len(train_df)} | Val size: {len(val_df)} | Test size: {len(test_df)}")
```
