# Notebook 03: Multi-Method Explainable AI (XAI) Analysis
## Attention Maps, SHAP, LIME, and Integrated Gradients (Captum)

This notebook demonstrates token-level interpretability on medical symptom text using four distinct XAI techniques.

```python
from src.inference.predict import MedicalSymptomClassifier
from src.explainability.token_importance import XAIEngine

# Load fine-tuned BioBERT model
classifier = MedicalSymptomClassifier(model_dir='../models/biobert', config_path='../config.yaml')
sample_symptom = "I have high fever, severe headache, shivering with chills, and intense joint pain."

# Predict disease
result = classifier.predict(sample_symptom)
print("Predicted Disease:", result['primary_prediction'])
print("Top-3 Predictions:", result['top_k_predictions'])
```

```python
# Run Multi-Method XAI Engine
xai = XAIEngine(classifier.model, classifier.tokenizer, classifier.label_list, classifier.device)
target_id = result['top_k_predictions'][0]['class_id']
xai_outputs = xai.explain_all(sample_symptom, target_class_id=target_id)

# View Summary DataFrame
xai_outputs['summary_df']
```
